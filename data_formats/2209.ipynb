```python
import csv 

users = [
    ["id" , "name" , "age" ],
    [1, "Abhay" , 23],
    [2 , "Amit" , 25],
    [3 , "suresh", 28]
]

with open("data/new_users.csv" , "w", newline="") as file:
    write = csv.writer(file)

    write.writerows(users)
```

```python
import csv 

users = [
    {'id': 1, 'name': 'abhay', 'age':23},
    {'id':2 , 'name': 'amit', 'age':25},
]

with open("data/users2.csv",'w', newline="") as file:

    fieldnames = ['id', 'name', 'age']

    writer = csv.DictWriter(file,
                            fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(users)
```

```python
data = [
  {
    "id": 1,
    "name": "Abhay",
    "age": 23,
    "skills": ["Python", "SQL", "Docker"]
  },
  {
    "id": 2,
    "name": "Amit",
    "age": 25,
    "skills": ["Python", "PostgreSQL"]
  }
]

with open('data/usersw.json','w') as file:
    users = json.dumps(data)
    file.write(users)
    print(users)
```

```python
import pandas as pd

data = {
    "id": [1,2,3],
    "name": ["Abhay", "Amit", "suresh"],
    "age": [23, 25, 28],
    "salary": [50000, 60000, 70000]
}

df = pd.DataFrame(data)

print(df)
df.to_parquet("data/users.parquet")
```
